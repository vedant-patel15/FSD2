```python
df=pd.DataFrame({'Physics':[60,70,80],
                'Chemistry':[65,75,85],
                'Maths':[70,80,90]})
```

```python
result=df.apply(sum)
result
```

```python
result=df.apply(sum,axis=1)
result
```

```python
result=df.apply(lambda x:x.mean())
result
```

```python
df['total']=df.apply(lambda x:x.sum(),axis=1)
df
```

```python
def grade(total):
    if total>=250:
        return 'A'
    elif total>=200:
        return 'B'
    else:
        return 'C'  
```

```python
df['grade']=df['total'].apply(grade)
df
```

```python
df=pd.DataFrame({'width':[1,2],
                'height':[3,4]})
df
```

```python
def get_stats(row):
    return [row['width']*row['height'],2*(row['width']+row['height'])]
```

```python
df.apply(get_stats,axis=1,result_type='expand')
```

```python
df.apply(get_stats,axis=1,result_type='reduce')
```

```python
df.apply(get_stats,axis=1,result_type='broadcast')
```

```python
df=pd.DataFrame({'A':[1,2],'B':[10,20]})
df
```

```python
def add_values(x,y,z):
    return x+y+z
def add_values1(x,m):
    return x*m
```

```python
df1=df.apply(add_values,args=(1,2))
df1
```
