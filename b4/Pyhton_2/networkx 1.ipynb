```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python

G=nx.DiGraph()
# G=nx.Graph()

G.add_nodes_from(range(4))
l=[[0,1,0,1],[0,0,0,1],[1,0,0,0],[1,0,1,0]]

for i in range(4):
    for j in range(4):
        if l[i][j]==1:
            G.add_edge(i,j)
            
nx.draw_networkx(G,node_size=850,node_color='red',edge_color='blue',width=3)
plt.show()
```

```python
G=nx.Graph()

G.add_nodes_from(range(1,8))

G.add_edges_from([(1,7),(1,2),(2,5),(1,6),(2,3),(3,3),(7,4)])

nx.draw_networkx(G,node_size=500,node_color="purple",edge_color='black',width=3)
```
