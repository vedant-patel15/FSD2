```python
import pandas as pd
```

```python
df=pd.DataFrame({
    'Product Category':['Electronics','Furniture','Electronics','Furniture'],
    'sales':[200,150,300,100]
})
df
```

```python
result=df.groupby('Product Category',axis=0,as_index=True,sort=True)['sales'].count()
result
```

```python
result=df.groupby('Product Category',axis=0,as_index=False,sort=True)['sales'].count()
result
```

```python
df=pd.DataFrame({'Region':['East','East','West','West','East','West'],
                'Sales':[100,150,200,120,110,130],
                'Manager':['A','B','C','D','E','F']})
df
```

```python
def sales_range(group):
    return group['Sales'].max() - group['Sales'].min()
sales_data=df.groupby('Region').apply(sales_range)
print(sales_data)
```

```python
df=pd.DataFrame({'dept':['Sales','Sales','HR','HR','IT','IT'],
                'Salary':[50000,60000,40000,45000,70000,75000],
                'Bonus':[5000,6000,2000,2500,8000,8500]})
df
```

```python
result=df.groupby('dept').agg('mean')
result
```

```python
result=df.groupby('dept').agg(['mean','median','max'])
result
```

```python
result=df.groupby('dept').agg({'Salary':['mean','max'],'Bonus':'sum'})
result
```

```python
result=df.groupby('dept').agg(MeanSalary=('Salary','mean'),MaxSalary=('Salary','max'),TotalBonus=('Bonus','sum'))
result
```
