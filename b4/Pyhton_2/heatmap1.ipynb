```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data=np.random.randint(low=1,high=100,size=(10,10))
print(data)
sns.heatmap(data)
plt.show()
```

```python
data=np.random.randint(low=1,high=100,size=(10,10))
# print(data)
sns.heatmap(data,vmax=70,vmin=30)
plt.show()
```

```python
data=np.random.randint(low=1,high=100,size=(10,10))
# print(data)
sns.heatmap(data,cmap='tab20')
plt.show()
```

```python
data=np.random.randint(low=1,high=100,size=(10,10))
# print(data)
sns.heatmap(data,annot=True)
plt.show()
```

```python
data=np.random.randint(low=1,high=100,size=(10,10))
# print(data)
sns.heatmap(data,annot=True,vmax=70,vmin=30,cmap='Pastel1',linewidth=1,linecolor='black',cbar=True,xticklabels=False,yticklabels=False)
plt.show()
```

```python
df=pd.DataFrame(np.random.random((5,5)),columns=['a','b','c','d','e'])
print(df)

sns.heatmap(df)
plt.show()
```

```python
data=np.random.randint(low=1,high=100,size=(3,10))
# print(data)
plt.figure(figsize=(15,5))
sns.heatmap(data,annot=True,cmap='Pastel1',linewidth=2,linecolor='black',cbar=True)
plt.show()
```
